Clustering adalah teknik dalam machine learning yang digunakan untuk mengelompokkan data ke dalam beberapa grup berdasarkan kemiripan atau kedekatan karakteristik. Metode ini termasuk dalam unsupervised learning karena tidak memerlukan label untuk memisahkan data.

Clustering memiliki berbagai aplikasi seperti segmentasi pelanggan, analisis pola, pengenalan gambar, dan pengelompokan dokumen. Dalam clustering, tujuan utama adalah meminimalkan jarak antar data dalam satu kelompok (intra-cluster) dan memaksimalkan jarak antar kelompok (inter-cluster).

# **K-Means Clustering**

K-Means clustering adalah teknik yang digunakan untuk mengatur data ke dalam kelompok berdasarkan kesamaannya. Misalnya, online shop menggunakan K-Means untuk mengelompokkan pelanggan berdasarkan frekuensi pembelian dan pengeluaran, dengan menciptakan segmen seperti ‘Pembeli Hemat’, ‘Pembeli Sering’, dan ‘Pembeli Besar’ untuk pemasaran yang dipersonalisasi.

Algoritma ini bekerja dengan terlebih dahulu memilih beberapa titik pusat yang disebut centroid secara acak dan setiap titik data kemudian ditetapkan ke centroid terdekat yang membentuk klaster. Setelah semua titik ditetapkan ke klaster, centroid diperbarui dengan menemukan posisi rata-rata titik di setiap klaster. Proses ini berulang hingga centroid berhenti berubah dan membentuk klaster. Tujuan pengelompokan adalah untuk membagi titik data ke dalam klaster sehingga titik data yang serupa termasuk dalam kelompok yang sama.

k_means_clustering.webp

## **1.1 Langkah Kerja Algoritma K-Means**

- Tentukan jumlah cluster (K): Kita tentukan berapa banyak grup yang ingin dibuat. Misalnya, kita ingin mengelompokkan pelanggan menjadi 3 kelompok, maka K = 3.
-  Inisialisasi Centroid secara acak: Centroid adalah titik pusat setiap kelompok. Kita pilih beberapa titik acak sebagai centroid pertama kali.
- Kelompokkan data: Setiap titik data (misalnya, pelanggan) akan dikelompokkan ke cluster yang centroid-nya paling dekat. Daya tariknya adalah bahwa titik data yang lebih mirip akan dikelompokkan bersama.
- Update centroid: Setelah semua data dikelompokkan, kita perbarui posisi centroid dengan cara menghitung rata-rata posisi data dalam kelompok tersebut.
- Ulangi langkah 3 dan 4: Proses ini diulang beberapa kali hingga posisi centroid tidak berubah lagi atau kita mencapai batas iterasi yang telah ditentukan.


## **1.2 Menentukan Nilai K yang Tepat**

Meskipun K-Means bekerja dengan baik pada dataset training, penting untuk diingat bahwa kelemahan K-Means adalah kita harus menentukan jumlah cluster K, sebelum kita tahu bahwa nilai K sudah optimal. Jumlah cluster yang akan dipilih tidak selalu jelas dalam aplikasi dunia nyata, terutama jika kita bekerja dengan dataset berdimensi tinggi.

**a. Metode Siku (Elbow)**

Metode ini adalah sebuah pendekatan yang digunakan dalam analisis clustering untuk menentukan jumlah optimal dari cluster dalam data. Metode ini mencari titik dimana penurunan inersia mulai melambat, membentuk grafik yang menyerupai bentuk siku. JIka terdapat sudut atau penurunan dalam grafik, maka nilai cluster tersebut dapat dianggap yang terbaik.

**b. Metode Silhouette Score**

Silhouette score adalah metrik yang digunakan untuk mengevaluasi seberapa baik hasil clustering  dalam clustering data. Skor ini berkisar dari -1 hingga 1, di mana nilai yang lebih tinggi menunjukkan hasil clustering yang lebih baik. Nilai dihitung dengan mengukur kemiripan setiap titik data dengan cluster yang dimilikinya dan seberapa berbedangan dengan cluster lain.

a = jarak rata-rata ke data lain dalam cluster yang sama.

b = jarak rata-rata ke data dalam cluster terdekat lain.

| Skor Silhouette | Kategori |
| --- | --- |
| 0,7 < skor ≤  1 | Strong Structure |
| 0,5 < skor ≤  0,7 | Medium Structure |
| 0,25 < skor ≤  0,5 | Weak Structure |
| skor ≤  0,25 | No Structure |



**c. Metode Gap Statistics**

Gap Statistic merupakan metode untuk menduga kelompok optimum pada analisis cluster. Gap ini dihasilkan dari membandingkan variasi antara data asli dan data acak. Nilai gap yang tinggi menunjukkan jumlah cluster yang optimal. Hal ini membantu kita memahami apakah pola yang kita lihat dalam data benar-benar bermakna atau hanya kebetulan.

# **DBSCAN Clustering**

Density-Based Spatial Clustering of Applications with Noise (DBSCAN) adalah algoritma clustering berbasis densitas yang mengelompokkan titik-titik data yang berdekatan dan menandai outlier sebagai noise berdasarkan kepadatannya di ruang fitur.  **DBSCAN mengelompokkan data berdasarkan seberapa padat titik-titik itu berada satu sama lain.**

Tidak seperti K-Means atau clustering yang bersifat hierarkis lain yang mengasumsikan cluster bersifat kompak dan bulat, DBSCAN unggul dalam menangani ketidakteraturan data di dunia nyata seperti:


- Cluster berbentuk sembarangan: Cluster dapat berbentuk apa saja, tidak hanya melingkar atau cembung.
- Noise dan Outlier: Secara efektif mengidentifikasi dan menangani noise points tanpa melemparnya ke cluster lain.

Databases.webp

**Langkah Kerja Algoritma DBSCAN:**
1. Tentukan parameter eps (radius) dan min_samples (jumlah minimum poin dalam radius).
2. Mulai dengan titik data awal, cek apakah jumlah titik dalam radius eps memenuhi min_samples.
- Jika ya, jadikan titik ini sebagai titik inti dan bentuk cluster.
- Jika tidak, tandai sebagai noise atau outlier.
3. Ulangi proses untuk semua titik sampai tidak ada lagi titik yang dapat dimasukkan ke cluster.


Keterangan:

1. eps (epsilon): Jarak maksimum antara dua titik untuk dianggap sebagai "tetangga dekat".

2. minPts: Jumlah minimum tetangga (termasuk dirinya sendiri) agar suatu titik disebut core point.

3. Core Point: Titik yang punya cukup tetangga (≥ minPts).

4. Border Point: Titik yang berada dalam jangkauan eps dari core point, tapi jumlah tetangganya < minPts.

5. Noise Point (Outlier): Titik yang bukan core dan bukan border — dianggap tidak masuk cluster mana pun.

### (C). Gap Statistic

```python
def gap_statistic(X, n_refs=5, max_k=10, random_state=42):
    np.random.seed(random_state)
    gaps = np.zeros(max_k)
    sdk = np.zeros(max_k)
    ref_disps = np.zeros((n_refs, max_k))

    for k in range(1, max_k+1):
        km = KMeans(n_clusters=k, random_state=random_state)
        km.fit(X)
        intra_dists = np.mean(np.min(cdist(X, km.cluster_centers_, 'euclidean'), axis=1))

        for i in range(n_refs):
            random_ref = np.random.random_sample(size=X.shape)
            km.fit(random_ref)
            ref_disps[i, k-1] = np.mean(np.min(cdist(random_ref, km.cluster_centers_, 'euclidean'), axis=1))

        gap = np.log(np.mean(ref_disps[:, k-1])) - np.log(intra_dists)
        sdk_val = np.sqrt(np.mean((np.log(ref_disps[:, k-1]) - np.log(np.mean(ref_disps[:, k-1])))**2)) * np.sqrt(1 + 1/n_refs)
        gaps[k-1] = gap
        sdk[k-1] = sdk_val

    return gaps, sdk

gaps, sdk = gap_statistic(df_scaled.values, n_refs=5, max_k=10)
```

```python
optimal_k = 1
for k in range(len(gaps)-1):
    if gaps[k] >= gaps[k+1] - sdk[k+1]:
        optimal_k = k + 1
        break

print(f'Jumlah cluster optimal berdasarkan Gap Statistic adalah: {optimal_k}')
```

## **DBSCAN CLUSTERING**

```python
min_samples = 5

X = df_scaled.drop(columns=['dbscan_cluster'], errors='ignore')

neighbors = NearestNeighbors(n_neighbors=min_samples)
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)

k_distances = np.sort(distances[:, min_samples - 1])
```

```python
plt.figure(figsize=(8, 4))
plt.scatter(range(len(k_distances)), k_distances, s=10, color='blue')
plt.ylabel(f'Jarak ke Tetangga ke-{min_samples}')
plt.xlabel('Data Point diurutkan berdasarkan jarak')
plt.title('Grafik Jarak K-Tetangga untuk Menentukan Parameter eps (Scatter Plot)')
plt.grid(True)
plt.show()
```

```python
print("Pada grafik jarak K-tetangga yang telah dibuat, kita melihat adanya 'elbow' atau titik perubahan kemiringan yang cukup signifikan di sekitar jarak 2,5 sampai 3,0.")
```

```python
eps = (2.5 + 3.0) / 2  # nilai tengah dari 2.5 dan 3.0

plt.figure(figsize=(8, 4))
plt.scatter(range(len(k_distances)), k_distances, s=10, color='blue')
plt.axhline(y=eps, color='red', linestyle='--', label=f'eps = {eps:.2f}')  # garis eps
plt.ylabel(f'Jarak ke Tetangga ke-{min_samples}')
plt.xlabel('Data Point diurutkan berdasarkan jarak')
plt.title('Grafik Jarak K-Tetangga untuk Menentukan Parameter eps (Scatter Plot)')
plt.legend()
plt.grid(True)
plt.show()

print(f'Parameter eps yang ditetapkan: {eps}')
```

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

dbscan = DBSCAN(eps=eps, min_samples=min_samples)
df_scaled['dbscan_cluster'] = dbscan.fit_predict(X)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X.iloc[:, 0],
    y=X.iloc[:, 1],
    hue=df_scaled['dbscan_cluster'],
    palette='Set1',
    style=df_scaled['dbscan_cluster'],
    legend='full',
    s=40
)

plt.title('Visualisasi Hasil Clustering DBSCAN')
plt.xlabel('Fitur 1')
plt.ylabel('Fitur 2')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```
