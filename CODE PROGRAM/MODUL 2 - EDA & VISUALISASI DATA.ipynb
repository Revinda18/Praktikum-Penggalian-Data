### Handling Missing Value

 Missing value merupakan kejadian yang umum terjadi ketika mengolah data, dikarenakan data yang tidak tersedia sehingga terdapat beberapa data yang hilang. Missing value dapat dideteksi dengan munculnya nilai NaN (Not Available) atau blank space pada suatu tabel. Mari kita coba.


````python
df.isnull() # Memeriksa data NaN
df.isna()
````

Dapat kita lihat dari output di atas, apabila value adalah False, maka data tersebut tidak NaN, begitu juga sebaliknya. Apabila value dari suatu data adalah True, maka data tersebut NaN atau missing value.

Untuk menampilkan jumlah data NaN, kita akan menambahkan syntax **sum()** seperti contoh berikut. Angka pada masing - masing kolom menampilkan jumlah missing value pada kolom dataset.


````python
df.isnull().sum() # Menghitung jumlah data NaN
````

Dari output di atas, kita mendapat informasi bahwa beberapa kolom memiliki data NaN/missing value. Beberapa cara yang dapat kita lakukan adalah sebagai berikut.

a. Menghilangkan atau menghapus data

`df.dropna(axis=0 , inplace=True)`

Fungsi **dropna()** akan menghapus data yang memiliki nilai NaN. Parameter axis=0 akan menentukan bahwa baris dengan nilai NaN akan dihapus, sedangkan axis=1 akan menghapus kolom yang memiliki nilai NaN. Selanjutnya, parameter inplace=True memastikan bahwa perubahan dilakukan langsung pada DataFrame asli tanpa perlu menyimpan hasilnya ke variabel baru, sementara nilai default inplace=False hanya akan membuat salinan DataFrame yang telah dimodifikasi tanpa mengubah DataFrame asli.

## Handle Outliers

Outlier merupakan suatu observasi pada kumpulan data yang memiliki pola yang berbeda dari observasi-observasi lainnya pada kumpulan data tersebut. Artinya, terdapat nilai yang jauh lebih kecil atau jauh lebih besar dibandingkan data lainnya dalam dataset. Penyebab munculnya outlier antara lain:

*   Adanya kesalahan prosedur dalam menginputkan data
*   Kesalahan dalam pengukuran atau analisis
*   Adanya keadaan khusus
Outlier


Lalu, mengapa penting untuk mengidentifikasi outliers?
Outliers dapat memengaruhi mean pada data dan membuat hasil analisis tidak akurat. Selain itu, kita dapat memeriksa apakah ada kesalahan pada data yang perlu diperbaiki. Kadang, outliers adalah data penting yang mewakili situasi ekstrem (contoh: makanan sangat tinggi kalori).

Berikut merupakan beberapa cara mendeteksi outlier:

### Boxplot

Boxplot adalah metode visual yang efektif untuk mengidentifikasi outliers. Outliers seringkali ditampilkan sebagai titik-titik yang berada di luar "whiskers" boxplot.

Mari kita coba untuk menampilkan outliers pada salah satu kolom DataFrame

````python
import matplotlib.pyplot as plt

plt.boxplot(df['calories'])
plt.title('Boxplot of Calories')
plt.ylabel('Calories')
plt.show()
````

Hasil visualisasi di atas dapat kita analisa sebagai berikut :
1. Kotak tengah (box) mewakilio data antara kuartil pertama (Q1) dan kuartil 3 (Q3). Sedangkan garis di dalam kotak adalah median. Data calories menunjukkan mayoritas nilai kalori berada antara 250 hingga sekitar 750.
2. Whiskers (Garis Horizontal) menunjukkan rentang data "normal" tanpa outliers. Data calories menunjukkan batas atas data normal ada di sekitar 1000.
3. Titik di atas atau bawah whiskers merupakan **outliers**. Titik - titik ini berada di luar batas normal (lebih kecil dari lower bound atau lebih besar dari upper bound). Pada visualisasi di atas, outliers berada diatas 1000. Hal ini berarti bahwa ada makanan dengan kalori lebih dari 1500 hingga 2500, yang jauh lebih besar dari data lainnya.

### Z-score

Z-Score mengukur seberapa jauh suatu pengamatan dari mean, dalam satuan standar deviasi. Dengan kata lain, z-score menunjukkan apakah sebuah nilai itu:

- Di atas rata-rata (z-score positif).
- Di bawah rata-rata (z-score negatif).
- Sama dengan rata-rata (z-score = 0)

Dengan menggunakan Z-Score, kita dapat mendeteksi outliers suatu data. Data dengan z > 3 atau z < -3 sering dianggap sebagai outliers. Z-Score mengubah data menjadi distribusi dengan rata-rata 0 dan standar deviasi 1, sehingga cocok untuk algoritma *machine learning*.

````python
from scipy import stats

z_scores = np.abs(stats.zscore(df['calories']))
outliers = np.where(z_scores > 3)

print("Indeks outliers:", outliers)
````

### Interquartile Range (IQR)

Cara lain untuk mendeteksi outliers adalah dengan menggunakan Interquartile Range (IQR). IQR fokus ke kuartil data, yaitu pembagian data menjadi empat bagian yang sama besar



> IQR = Q3 - Q1
* Q1 (Kuartil Pertama): Nilai di mana 25% data berada di bawahnya.
- Q3 (Kuartil Ketiga): Nilai di mana 75% data berada di bawahnya.
- IQR: Jarak antara Q3 dan Q1, menunjukkan rentang data utama.

Kita dapat menggunakan IQR untuk mendeteksi outliers apabila data memiliki distribusi yang tidak normal. IQR efektif untuk mendeteksi data yang jauh di luar rentang utama.



````python
Q1 = np.percentile(df['calories'], 25)
Q3 = np.percentile(df['calories'], 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = np.where((df['calories'] < lower_bound) | (df['calories'] > upper_bound))
print("Indeks outliers:", outliers)
````

## Data Visualization

Ada banyak sekali jenis visualisasi data, dari yang sederhana hingga yang kompleks. Setiap jenis visualisasi biasanya lebih tepat digunakan untuk suatu tujuan tertentu. Secara garis besar, ada empat tujuan visualisasi secara umum.

### Bar Chart

### Multi Bar Chart

````python
import matplotlib.pyplot as plt
import numpy as np

multi_data_bar = {'Bar1': [20, 35, 30, 35, 27],
                  'Bar2': [25, 32, 34, 20, 25]}

plt.figure(figsize=(13, 9))  # Menentukan ukuran visualisasi
plt.subplot(2, 2, 2)  # row 2, col 2, posisi 2
bar_width = 0.35
index = np.arange(len(multi_data_bar['Bar1']))

plt.bar(index, multi_data_bar['Bar1'], bar_width, label='Bar1')
plt.bar(index + bar_width, multi_data_bar['Bar2'], bar_width, label='Bar2')
plt.title('Multi Bar Chart')
plt.legend()
plt.show()
````

### Line Chart

````python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]


plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='Prime Numbers', color='green',
         marker='o', linestyle='--')
plt.plot(x, y2, label='Pattern Numbers', color='red',
         marker='s')
plt.title('Line Chart', fontsize=16)
plt.xlabel('X-axis', fontsize=12)
plt.ylabel('Y-axis', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=10)


plt.show()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = '/content/music_tours.csv'
music_tours_data = pd.read_csv(file_path)

def clean_and_convert_to_float(series):
    return series.replace({'\$': '', ',': '', '[a-zA-Z\[\]]': ''}, regex=True).astype(float)

music_tours_data['Actual gross'] = clean_and_convert_to_float(music_tours_data['Actual gross'])
music_tours_data['Adjusted gross (in 2022 dollars)'] = clean_and_convert_to_float(music_tours_data['Adjusted gross (in 2022 dollars)'])
music_tours_data['Average gross'] = clean_and_convert_to_float(music_tours_data['Average gross'])

music_tours_data['Shows'] = music_tours_data['Shows'].astype(int)

descriptive_stats = music_tours_data[['Actual gross', 'Adjusted gross (in 2022 dollars)', 'Average gross', 'Shows']].describe()

print("Descriptive Statistics:\n", descriptive_stats)

plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.boxplot(music_tours_data['Actual gross'])
plt.title('Boxplot: Actual Gross Revenue')

plt.subplot(1, 3, 2)
plt.boxplot(music_tours_data['Adjusted gross (in 2022 dollars)'])
plt.title('Boxplot: Adjusted Gross Revenue')

plt.subplot(1, 3, 3)
plt.boxplot(music_tours_data['Average gross'])
plt.title('Boxplot: Average Gross Revenue')

plt.tight_layout()
plt.show()

def detect_outliers_IQR(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series[(series < lower_bound) | (series > upper_bound)]

outliers_actual_gross = detect_outliers_IQR(music_tours_data['Actual gross'])
outliers_adjusted_gross = detect_outliers_IQR(music_tours_data['Adjusted gross (in 2022 dollars)'])

print("\nOutliers in Actual Gross Revenue:\n", outliers_actual_gross)
print("\nOutliers in Adjusted Gross Revenue:\n", outliers_adjusted_gross)

````
